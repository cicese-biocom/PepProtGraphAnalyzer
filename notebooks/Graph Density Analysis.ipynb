```python
import pandas as pd
from plotnine.data import mpg
from plotnine import ggplot, aes, geom_histogram, geom_bar
from plotnine.exceptions import PlotnineWarning
import warnings
warnings.filterwarnings("ignore", category=PlotnineWarning)
```

```python
def plot_inter_amino_acid_distances(inter_amino_acid_distances):
    graphic = (
        ggplot(inter_amino_acid_distances, aes(x="inter_amino_acid_distances")) 
        + geom_histogram(binwidth=0.001, fill='blue', color='black')  
    )
    graphic.save(f'Inter_Amino_Acid_Distances-{distance_function}.png', dpi=300)
    return graphic
```

```python
distance_function = 'angular_separation'
filename = f'data/Graph_Analysis/Inter_Amino_Acid_Distances-{distance_function}.csv'
inter_amino_acid_distances = pd.read_csv(filename, header=None, names=['inter_amino_acid_distances'])
plot_inter_amino_acid_distances(inter_amino_acid_distances)
```
