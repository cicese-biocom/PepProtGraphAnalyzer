### **Sequence Length Data**

### **Analysis by Interval Length**

```python
intervals = {
    '[1-10)': (1, 9),
    '[10-20)': (10, 19),
    '[20-30)': (20, 29),
    '[30-40)': (30, 39),
    '[40-50)': (40, 49),
    '[50-60)': (50, 59),
    '[60-70)': (60, 69),
    '[70-80)': (70, 79),
    '[80-90)': (80, 89),
    '[90-100]': (90, 100),
    '>100': (101, float('inf'))
}

summary = {key: ((data['length'] >= val[0]) & (data['length'] <= val[1])).sum() for key, val in intervals.items()}

output = paths['sequence_length_path'].joinpath("Sequence Length Summary.csv")

summary_df = pd.DataFrame.from_dict(summary, orient='index', columns=['Count'])
summary_df.to_csv(output)
summary_df
```
